## Setup — Logger
Run this cell once at the start of every session. It initialises the log file and exposes `log_step()` / `update_step()` helpers used by all phase cells.

````python
import json
import os
from datetime import datetime
from pathlib import Path
from IPython.display import display, HTML

# ── paths ──────────────────────────────────────────────────────────────────────
NOTEBOOK_DIR = Path(globals().get('__vsc_ipynb_file__', __file__ if '__file__' in dir() else '')).parent \
    if '__vsc_ipynb_file__' in globals() else Path.cwd()
LOG_PATH = NOTEBOOK_DIR / 'fazdane_build_log.json'

# ── load or initialise log ─────────────────────────────────────────────────────
def _load_log():
    if LOG_PATH.exists():
        with open(LOG_PATH, 'r') as f:
            return json.load(f)
    return {"project": "FazDane Databricks Finance Platform",
            "created": datetime.now().isoformat(),
            "steps": []}

def _save_log(data):
    with open(LOG_PATH, 'w') as f:
        json.dump(data, f, indent=2)

# ── public helpers ─────────────────────────────────────────────────────────────
def log_step(phase: int, task: str, status: str = 'in_progress', notes: str = ''):
    """Append a new step to the log."""
    data = _load_log()
    entry = {
        "id": len(data['steps']) + 1,
        "phase": phase,
        "task": task,
        "status": status,          # in_progress | complete | skipped | blocked
        "started": datetime.now().isoformat(),
        "completed": None,
        "notes": notes
    }
    data['steps'].append(entry)
    _save_log(data)
    print(f"[LOG] Phase {phase} | {task} → {status}")
    return entry['id']

def update_step(step_id: int, status: str, notes: str = ''):
    """Update an existing step (mark complete, add notes, etc.)."""
    data = _load_log()
    for s in data['steps']:
        if s['id'] == step_id:
            s['status'] = status
            s['notes'] = notes if notes else s['notes']
            if status == 'complete':
                s['completed'] = datetime.now().isoformat()
            _save_log(data)
            print(f"[LOG] Step {step_id} updated → {status}")
            return
    print(f"[WARN] Step {step_id} not found in log.")

def show_log():
    """Render current log as a formatted HTML table."""
    data = _load_log()
    STATUS_COLOR = {
        'complete':    '#d4edda',
        'in_progress': '#fff3cd',
        'skipped':     '#e2e3e5',
        'blocked':     '#f8d7da',
        'not_started': '#f8f9fa'
    }
    rows = ''.join(
        f"<tr style='background:{STATUS_COLOR.get(s['status'],'#fff')}'>"
        f"<td>{s['id']}</td><td>{s['phase']}</td><td>{s['task']}</td>"
        f"<td><b>{s['status']}</b></td>"
        f"<td>{(s['started'] or '')[:19]}</td>"
        f"<td>{(s['completed'] or '')[:19]}</td>"
        f"<td>{s['notes']}</td></tr>"
        for s in data['steps']
    )
    html = f"""
    <h3>FazDane Build Log — {len(data['steps'])} steps</h3>
    <table border='1' cellpadding='5' style='border-collapse:collapse;font-size:13px;width:100%'>
      <thead style='background:#1F4E79;color:white'>
        <tr><th>#</th><th>Phase</th><th>Task</th><th>Status</th><th>Started</th><th>Completed</th><th>Notes</th></tr>
      </thead>
      <tbody>{rows}</tbody>
    </table>"""
    display(HTML(html))

print('Logger ready. Log file:', LOG_PATH)
````

## Progress Dashboard
Run the cell below at any time to see the full log with colour-coded status.

````python
# ── Progress Dashboard ─────────────────────────────────────────────────────────
import json
from pathlib import Path
from IPython.display import display, HTML

def show_summary():
    """Print phase-by-phase completion summary."""
    LOG_PATH = Path.cwd() / 'fazdane_build_log.json'
    if not LOG_PATH.exists():
        print('No log found. Run the Setup cell and Phase log cells first.')
        return
    with open(LOG_PATH) as f:
        data = json.load(f)
    steps = data['steps']
    phases = sorted(set(s['phase'] for s in steps))
    rows = ''
    total = len(steps)
    done  = sum(1 for s in steps if s['status'] == 'complete')
    for ph in phases:
        ph_steps = [s for s in steps if s['phase'] == ph]
        n_done   = sum(1 for s in ph_steps if s['status'] == 'complete')
        pct      = int(n_done / len(ph_steps) * 100) if ph_steps else 0
        bar_w    = pct
        bar      = f"<div style='background:#1F4E79;width:{bar_w}%;height:14px;border-radius:4px'></div>"
        rows    += (f"<tr><td>Phase {ph}</td><td>{n_done}/{len(ph_steps)}</td>"
                    f"<td><div style='background:#e9ecef;border-radius:4px;width:200px'>{bar}</div></td>"
                    f"<td>{pct}%</td></tr>")
    html = f"""
    <h3>FazDane Build Progress — {done}/{total} tasks complete</h3>
    <table border='1' cellpadding='6' style='border-collapse:collapse;font-size:13px'>
      <thead style='background:#1F4E79;color:white'>
        <tr><th>Phase</th><th>Tasks Done</th><th>Progress</th><th>%</th></tr>
      </thead><tbody>{rows}</tbody>
    </table>"""
    display(HTML(html))

show_summary()
````
